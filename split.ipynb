```python
import os
import shutil
```

For right now, we are only looking at the ones stored in the google folder 

```python
bigF = 'data/bdappv/google/img' 
smallF = 'data/bdappv/google/mask' #this is the folder that has detected solar planels in 
```

Create the following folders in the google folder 

```python
posF = 'data/bdappv/google/pos_img' 
negF = 'data/bdappv/google/neg_img' 
```

Split the google/img folder into positive and negative cases 

```python
big_folder_files = set(os.listdir(bigF))
small_folder_files = set(os.listdir(smallF))

# Iterate through files in the big folder
for file in big_folder_files:
    # Determine the target path based on whether the file exists in the small folder
    if file in small_folder_files:
        target_folder = posF
    else:
        target_folder = negF
    
    # Move the file to the respective folder
    shutil.move(os.path.join(bigF, file), os.path.join(target_folder, file))

print("Files have been sorted into positive and negative cases.")
```

```python
# Get the list of files in each directory
positive_files = os.listdir(posF)
negative_files = os.listdir(negF)

# Count the number of files
positive_count = len(positive_files)
negative_count = len(negative_files)

# Print the counts
print(f"Number of files in Positive Cases: {positive_count}")
print(f"Number of files in Negative Cases: {negative_count}")
```
